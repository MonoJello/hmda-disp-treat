```python
# Add the repeated ethnicity columns.
for person in ["applicant", "co_applicant"]:
    for number in range(1, 6):
        code_maps[f"{person}_ethnicity_{number}"] = ethnicity_map



# Add the repeated race columns.
for person in ["applicant", "co_applicant"]:
    for number in range(1, 6):
        code_maps[f"{person}_race_{number}"] = race_map

# Add sex columns.
code_maps["applicant_sex"] = sex_map
code_maps["co_applicant_sex"] = sex_map

# Add visual-observation/surname columns.
code_maps["applicant_ethnicity_observed"] = observed_map
code_maps["co_applicant_ethnicity_observed"] = observed_map
code_maps["applicant_race_observed"] = observed_map
code_maps["co_applicant_race_observed"] = observed_map
code_maps["applicant_sex_observed"] = observed_map
code_maps["co_applicant_sex_observed"] = observed_map

# Add the same denial-reason definitions to denial_reason_2 through denial_reason_4.
for number in range(2, 5):
    code_maps[f"denial_reason_{number}"] = code_maps["denial_reason_1"]

# Add Automated Underwriting System definitions.
aus_map = {
    "1": "Desktop Underwriter (DU)",
    "2": "Loan Prospector (LP)",
    "3": "Technology Open to Approved Lenders (TOTAL) Scorecard",
    "4": "Guaranteed Underwriting System (GUS)",
    "5": "Other",
    "6": "Not applicable",
    "1111": "Not applicable",
}

for number in range(1, 6):
    code_maps[f"aus_{number}"] = aus_map
```

```python
for column, mapping in code_maps.items():
    if column in df.columns:
        values = df[column].astype("string").str.strip()

        # Add a new column such as applicant_ethnicity_1_definition
        # df[f"{column}_definition"] = values.map(mapping)
        df[f"{column}"] = values.map(mapping)

        # Report unexpected codes
        nonblank_values = values.dropna()
        unmapped = sorted(
            set(nonblank_values) - set(mapping.keys())
        )

        if unmapped:
            print(f"{column}: unmapped values: {unmapped}")
```

```python
mapped_columns = [
    column for column in code_maps
    if column in df.columns
]

print(f"Mapped columns: {len(mapped_columns)}")
print(mapped_columns)
```
