```python
import requests

years = 2025
leis = '7H6GLXDRUGQFU57RNE97'
```

```python
import requests

url = f"https://ffiec.cfpb.gov/file/modifiedLar/year/{years}/institution/{leis}/txt/header"
out_path = f"../data/{years}_{leis}.csv"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "text/csv",
}

r = requests.get(url, headers=headers, timeout=60)

print(r.status_code)
r.raise_for_status()

with open(out_path, "wb") as f:
    f.write(r.content)

print(f"Saved CSV to {out_path}")
```

```python
import hashlib

file_path = out_path

sha256 = hashlib.sha256()

with open(file_path, "rb") as file:
    for chunk in iter(lambda: file.read(1024 * 1024), b""):
        sha256.update(chunk)

file_hash = sha256.hexdigest()

print(file_hash)
```
